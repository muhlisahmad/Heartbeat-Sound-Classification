<h2>Augmentasi Data</h2>
<p style="text-align: justify"> Augmentasi data adalah proses dimana kami membuat sampel data sintetis baru dengan mendefinisikan fungsi untuk menambahkan dan mengubah detail-detail kecil pada dataset yang kami peroleh. Untuk menghasilkan data audio sintetis, kita dapat menerapkan penambahan gangguan (<i>noise addition</i>), pergeseran waktu (<i>time shifting</i>), perubahan nada (<i>pitch shifting</i>) dan kecepatan (<i>time stretching</i>). </p>
<p style="text-align: justify"> Tujuan dilakukannya hal tersebut adalah untuk menambah kevariatifan pada data training yang kita miliki dengan mengubah detail kecil dari data audio yang asli. Hal ini membantu model CNN dalam meningkatkan kempampuan belajarnya untuk menggeneralisasi data. Hal ini serupa dengan augmentasi data pada gambar dimana dat gambar dilakukan pembalikan (<i>mirroring</i>), pembesaran (<i>zooming</i>), transformasi rotasi (<i>rotating</i>), dan lain-lain. </p>

<ol>
    <h3><li>Noise Addition</li></h3>
    <p style="text-align: justify"> Proses ini melibatkan penambahan <i>noise</i> yaitu ke dalam sampel data asli. <i>Noise</i> dihasilkan dengan menambahkan sampel acak yang didistribusikan secara merata (<i>uniform</i>). </p>
    <h3><li>Time Shifting</li></h3>
    <p style="text-align: justify"> Disini kita menggeser gelombang dengan faktor acak yang terdistribusi secara merat (<i>uniform</i>). Hal ini akan menggeser gelombang ke kanan dengan faktor yang diberikan dalam sumbu waktu. Untuk menerapkan hal ini, kita telah menggunakan fungsi <code>roll</code> pada <code>numpy</code> untuk menghasilkan pergeseran waktu. </p>
    <h3><li>Time Stretching</li></h3>
    <p style="text-align: justify"> Proses mengubah kecepatan/durasi suara tanpa mempengaruhi nada suara. Hal ini dapat dicapai dengan menggunakan fungsi <code>time_stretch</code> pada library <code>librosa</code>. Fungsi <code>time_stretch</code> mengambil sampel gelombang dan faktor yang digunakan untuk meregangkan sebagai input. Kita mendefinisikan set default untuk faktor ini yakni 0,6. </p>
    <h3><li>Pitch Shifting</li></h3>
    <p style="text-align: justify"> Proses ini adalah implementasi dari penskalaan nada yang digunakan dalam alat musik. Proses ini mengubah nada suara tanpa mempengaruhi kecepatannya. Kali ini kita akan menggunakan fungsi <code>pitch_shift</code> pada library <code>librosa</code>. fungsi ini membutuhkan sampel gelombang, <i>sample rate</i> dan <i>pitch factor</i>. Kami menetapkan default untuk <i>pitch factor</i> sebesar 0,7. </p>
</ol>

```python
def noise(data):
    noise_value = 0.009*np.random.uniform()*np.amax(data)
    data = data + noise_value*np.random.normal(size=data.shape[0])
    return data

def stretch(data, rate=0.6):
    return librosa.effects.time_stretch(data, rate)

def shift(data, samplingrate):
    return np.roll(data,samplingrate/10)

def pitch(data, samplingrate, pitch_factor=0.7):
    return librosa.effects.pitch_shift(data, samplingrate, pitch_factor)
```

<h2>Ekstraksi Fitur</h2>
<p style="text-align: justify"> Ekstraksi fitur merupakan bagian yang sangat penting dalam menganalisis dan menemukan hubungan antara fitur satu dengan yang lain. Seperti yang telah diketahui bahwa data audio yang ada tidak dapat dipahami oleh model secara langsung sehingga kita perlu mengubahnya menjadi format yang dapat dimengerti dengan melakukan ekstraksi fitur. </p>
<ol>
    <h3><li>Zero Crossing Rate</li></h3>
    <p style="text-align: justify"> Tingkat perubahan sinyal dalam melewati sumbu horizontal selama durasi <i>frame</i> tertentu. </p>
    <h3><li>MFCCs - Mel Frequency Cepstral Coefficients</li></h3>
    <p style="text-align: justify"> MFCCs Mel Frequency Cepstral Coefficients membentuk representasi cepstral dimana pita frekuensi tidak linier tetapi didistribusikan menurut skala mel. </p>
    <h3><li>Root Mean Square Energy</li></h3>
    <p style="text-align: justify"> Root Mean Square Energy didasarkan pada semua sampel dalam sebuah <i>frame</i>. Hal ini bertindak sebagai indikator kenyaringan, karena semakin tinggi energinya, semakin keras suaranya. Namun kurang sensitif terhadap <i>outlier</i> dibandingkan dengan Amplitudo Amplop. Fitur ini berguna dalam segmentasi audio dan klasifikasi genre musik. </p>
    <h3><li>Mel Spektogram</li></h3>
    <p style="text-align: justify"> Mel Spektogram secara logaritmik membuat frekuensi di atas ambang tertentu (frekuensi sudut). Skala mel adalah skala nada yang umumnya dirasakan oleh pendengaran manusia. Saat frekuensi meningkat, interval, dalam hertz, antara nilai skala mel (atau hanya mel) meningkat. Nama mel berasal dari melodi dan menunjukkan bahwa skala didasarkan pada perbandingan antara nada. </p>
</ol>

```python
def extract_features(data, samplerate):
    # ZCR
    result = np.array([])
    zcr = np.mean(librosa.feature.zero_crossing_rate(y=data).T, axis=0)
    result=np.hstack((result, zcr)) # stacking horizontally

    # Chroma_stft
    stft = np.abs(librosa.stft(data))
    chroma_stft = np.mean(librosa.feature.chroma_stft(S=stft, sr=samplerate).T, axis=0)
    result = np.hstack((result, chroma_stft)) # stacking horizontally

    # MFCC
    mfcc = np.mean(librosa.feature.mfcc(y=data, sr=samplerate).T, axis=0)
    result = np.hstack((result, mfcc)) # stacking horizontally

    # Root Mean Square Value
    rms = np.mean(librosa.feature.rms(y=data).T, axis=0)
    result = np.hstack((result, rms)) # stacking horizontally

    # MelSpectogram
    mel = np.mean(librosa.feature.melspectrogram(y=data, sr=samplerate).T, axis=0)
    result = np.hstack((result, mel)) # stacking horizontally
    
    return result

def get_features(path):
    # duration and offset are used to take care of the no audio in start and the ending of each audio files as seen above.
    data = librosa.load(path)[0]
    rate = wavfile.read(path)[0]
    
    # without augmentation
    res1 = extract_features(data, rate)
    result = np.array(res1)
    
    # data with noise
    noise_data = noise(data)
    res2 = extract_features(noise_data, rate)
    result = np.vstack((result, res2)) # stacking vertically
    
    # data with stretching and pitching
    new_data = stretch(data)
    data_stretch_pitch = pitch(new_data, rate)
    res3 = extract_features(data_stretch_pitch, rate)
    result = np.vstack((result, res3)) # stacking vertically
    
    return result
```

```python
X, Y = [], []
for path, label in zip(data_ab.fname, data_ab.label):
    feature = get_features(path)
    for element in feature:
        X.append(element)
        Y.append(label)
```

```python
Features = pd.DataFrame(X)
Features['label'] = Y
Features.to_csv('features.csv', index=False)
Features.head()
```

```python
print("Jumlah data fitur setiap label =",str(len(data_ab.label.unique())),"kategori, diantaranya:")

for i in Features.label.unique():
    print(f"- {i}\t: {Features['label'].value_counts()[i]}")
```

<h2>Data Separation</h2>
<p style="text-align: justify"> Setelah kita lakukan ekstraksi fitur pada data, sekarang kita perlu normalisasi dan membagi data kita untuk data pelatihan (<i>data train</i>) dan data pengujian (<i>data test</i>) </p>

```python
X = Features.iloc[:,:-1].values
Y = Features['label'].values
enc = OneHotEncoder()
scaler = StandardScaler()
Y_Encode = enc.fit_transform(np.array(Y).reshape(-1,1)).toarray()
```

```python
print(f"Ukuran data fitur : {X.shape}")
print(f"Ukuran data label : {Y_Encode.shape}")
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y_Encode, train_size=0.8, random_state=4, shuffle=True)
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
x_train = np.expand_dims(x_train,axis=2)
x_test = np.expand_dims(x_test,axis=2)
```

```python
print(f"Ukuran data train : {x_train.shape}")
print(f"Ukuran data test : {x_test.shape}")
```

```python
Early_Stopper = EarlyStopping(monitor="loss",patience=3,mode="min")
rlrp = ReduceLROnPlateau(monitor='val_loss', factor=0.4, verbose=0, patience=2, min_lr=0.001)
Checkpoint_Model = ModelCheckpoint(monitor="val_accuracy",
                                   mode='max',
                                   save_best_only=True,
                                   save_weights_only=True,
                                   filepath="./modelcheck")
```

<h2>Modelling</h2>

```python
model=Sequential()
model.add(Conv1D(256, kernel_size=5, strides=1, padding='same', activation='relu', input_shape=(x_train.shape[1], 1)))
model.add(MaxPooling1D(pool_size=3, strides = 2, padding = 'same'))

model.add(Conv1D(256, kernel_size=4, strides=1, padding='same', activation='relu'))
model.add(MaxPooling1D(pool_size=3, strides = 2, padding = 'same'))
model.add(Dropout(0.3))

model.add(Conv1D(128, kernel_size=4, strides=1, padding='same', activation='relu'))
model.add(MaxPooling1D(pool_size=3, strides = 2, padding = 'same'))
model.add(Dropout(0.3))

model.add(Conv1D(64, kernel_size=4, strides=1, padding='same', activation='relu'))
model.add(MaxPooling1D(pool_size=3, strides = 2, padding = 'same'))
model.add(Dropout(0.3))

model.add(Conv1D(32, kernel_size=4, strides=1, padding='same', activation='relu'))
model.add(MaxPooling1D(pool_size=3, strides = 2, padding = 'same'))
model.add(Dropout(0.3))

model.add(Flatten())
model.add(Dense(units=1024, activation='relu'))
model.add(Dropout(0.3))

model.add(Dense(units=5, activation='softmax'))
model.compile(optimizer = 'adam' , loss = 'categorical_crossentropy' , metrics = ['accuracy'])

model.summary()
```

<h2>Melatih Model</h2>
<p style="text-align: justify"> Disini kita secara eksplisit mengatur berapa lama kita menginginkan setiap <code>epoch</code> berjalan. Lalu jalankan sel berikut untuk melihat hasilnya. Pelatihan akan memakan waktu yang lama hal ini merupakan hal yang wajar mengingat kita melatih data yang cukup besar. </p>

```python
Conv1D_Model = model.fit(x_train, y_train, batch_size=12, epochs=50, validation_data=(x_test, y_test), callbacks=[rlrp, Checkpoint_Model])
```

<h2>Melakukan Plot Akurasi Model</h2>

```python
Model_Results = model.evaluate(x_test,y_test)
```

```python
print(f"LOSS\t\t: {round(Model_Results[0]*100,2)}%")
print(f"ACCURACY\t: {round(Model_Results[1]*100,2)}%")

epochs = [i for i in range(50)]
fig , ax = plt.subplots(1,2)
train_acc = Conv1D_Model.history['accuracy']
train_loss = Conv1D_Model.history['loss']
test_acc = Conv1D_Model.history['val_accuracy']
test_loss = Conv1D_Model.history['val_loss']

fig.set_size_inches(20,6)
ax[0].plot(epochs , train_loss , label = 'Training Loss')
ax[0].plot(epochs , test_loss , label = 'Testing Loss')
ax[0].set_title('Training & Testing Loss')
ax[0].legend()
ax[0].set_xlabel("Epochs")

ax[1].plot(epochs , train_acc , label = 'Training Accuracy')
ax[1].plot(epochs , test_acc , label = 'Testing Accuracy')
ax[1].set_title('Training & Testing Accuracy')
ax[1].legend()
ax[1].set_xlabel("Epochs")
plt.show()
```
